### 1. Initialize Environment & Dual Agents

```python
import os
from env import HouseholdEnvironment
from agents import DM_agent, ME_agent, MemoryEncoding, LLM_public

# HF_TOKEN is optional for public Hugging Face models such as Qwen/Qwen3-8B.
# Set it in your environment if your Hugging Face setup requires authentication.
llm_config = {
    "model_id": "Qwen/Qwen3-8B",
    "temperature": 0.0,
    "max_tokens": 1024,
    "dtype": "auto",
    "device": "auto",
    "enable_thinking": False,
}

# 1. Initialize environment
env = HouseholdEnvironment()

# 2. Load Qwen once and share it across both agents.
qwen_llm = LLM_public(**llm_config)
dm_agent = DM_agent(llm=qwen_llm)
me_agent = ME_agent(llm=qwen_llm)
memory_encoding = MemoryEncoding(me_agent=me_agent)
```

### 2. Run Dual-Agent Execution Loop & Save Messages/Video

```python
from module import run_agent_loop

# Run dual-agent interaction loop and export messages & execution video
results = run_agent_loop(
    env=env,
    dm_agent=dm_agent,
    memory_encoding=memory_encoding,
    debug=False,
    print_summary=False,
    display_retrieval_plots=True,
    save_video=True,
    save_chat=True,
    output_video_path="video_qwen.gif",
    output_messages_path="chat_qwen.json",
)
```
